```python
import numpy as np

from matplotlib import pyplot as plt
from scipy.stats import binom
```

```python
p_grid = np.linspace(0, 1, 1000)
lik = binom.pmf(8, 15, p_grid)
prior = np.ones((1000,))

post = lik*prior
post = post / np.sum(post)

plt.plot(
    p_grid,
    post
)

n_samples = int(1e4)
rng = np.random.default_rng(seed=1)
samples = rng.choice(p_grid, size=n_samples, p=post)

plt.hist(
    samples,
    bins = 30,
    density = False,
    edgecolor = "black"
)

plt.xlabel("Proportion (p)")
plt.ylabel("Frequency")
plt.title("Samples of proportion parameter from posterior")

```

```python
q1 = np.quantile(samples, q=0.05)
q2 = np.quantile(samples, q=0.95)

plt.plot(
    p_grid,
    post
)
plt.fill_between(
    p_grid, 
    post,
    where=(p_grid >= q1) & (p_grid <= q2),
    color="blue",
    alpha=0.4,
    label="95% PI"
)
```
